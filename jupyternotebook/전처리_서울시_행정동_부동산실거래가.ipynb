```python

cols=["CGG_CD","CGG_NM",
      "STDG_CD","STDG_NM",
      "CTRT_DAY",
      "THING_AMT",  # 만원
      "BLDG_USG"]

real_estate_df = pd.read_csv(real_estate_file, usecols=cols, encoding="utf-8-sig")

# 필요한 컬럼만
real_estate_df = real_estate_df[real_estate_df["BLDG_USG"]=="아파트"]

real_estate_df['법정동코드'] = real_estate_df["CGG_CD"] * 100000 + real_estate_df['STDG_CD']

print(f"갯수: {len(real_estate_df)}")
real_estate_df.head(2)

```

```python

# YYYYMM 추출
real_estate_df["YYYYMM"] = real_estate_df["CTRT_DAY"].astype(str).str[:6]

# 월별 집계
mm_real_estate_df = (
    real_estate_df.groupby(
        ["YYYYMM","법정동코드","STDG_NM"],
        as_index=False)
    .agg(
        THING_AMT_TOTAL=("THING_AMT", "sum"),
        THING_COUNT=("THING_AMT", "count")
    )
    .sort_values(
        ["YYYYMM", "법정동코드"]
    )
)

mm_real_estate_df['THING_AMT_AVG'] = (mm_real_estate_df['THING_AMT_TOTAL'] / mm_real_estate_df['THING_COUNT']).fillna(0).astype(int)

mm_real_estate_df.to_csv(
            "서울시_아파트실거래가_행정동_월별.csv",
            index=False,
            encoding="utf-8-sig")

print(f"갯수: {len(mm_real_estate_df)}")
mm_real_estate_df.head(2)


```

```python
# 분기별 행정동 평균 아파트 가격을 월별로  interpolate 한다.


mapping_df = pd.read_csv("서울시_행정동ID_행정동코드_맵핑.csv", encoding="utf-8-sig")

apt_price_file = "서울시 상권분석서비스(아파트-행정동).csv"
#apt_price_file = "서울시_부동산실거래가.csv"

apt_price_df = pd.read_csv(apt_price_file,
                           encoding="cp949"
                           #encoding="utf-8-sig"
                           ).sort_values(["기준_년분기_코드","행정동_코드"])


cols = [
    "기준_년분기_코드",
    "행정동_코드",
    #"행정동_코드_명",
    "아파트_단지_수",
    "아파트_면적_66_제곱미터_미만_세대_수",
    "아파트_면적_66_제곱미터_세대_수",
    "아파트_면적_99_제곱미터_세대_수",
    "아파트_면적_132_제곱미터_세대_수",
    "아파트_면적_165_제곱미터_세대_수",
    "아파트_평균_면적",
    "아파트_평균_시가"
]

apt_price_df[cols] = (
    apt_price_df[cols]
    .fillna(0)
    .astype(int)
)

value_cols = [
    "기준_년분기_코드",
    "행정동_코드",
    "행정동_코드_명",
    "아파트_단지_수",
    "아파트_면적_66_제곱미터_미만_세대_수",
    "아파트_면적_66_제곱미터_세대_수",
    "아파트_면적_99_제곱미터_세대_수",
    "아파트_면적_132_제곱미터_세대_수",
    "아파트_면적_165_제곱미터_세대_수",
    "아파트_평균_면적",
    "아파트_평균_시가"
]



apt_price_df = apt_price_df[apt_price_df["기준_년분기_코드"] >= 20242]

apt_price_df = apt_price_df[value_cols]

apt_price_df["행정동코드"] = apt_price_df["행정동_코드"] * 100

print(f"갯수: {len(apt_price_df)}")
apt_price_df.head(2)
```

```python
# interpolate를 통해서 월단위 데이터 전환

# 분기 -> 월 매핑
quarter_month_map = {
    "1": "03",
    "2": "06",
    "3": "09",
    "4": "12"
}

value_cols = [
    "아파트_단지_수",
    "아파트_면적_66_제곱미터_미만_세대_수",
    "아파트_면적_66_제곱미터_세대_수",
    "아파트_면적_99_제곱미터_세대_수",
    "아파트_면적_132_제곱미터_세대_수",
    "아파트_면적_165_제곱미터_세대_수",
    "아파트_평균_면적",
    "아파트_평균_시가"
]

code_cols = [
    "기준_년분기_코드",
    "행정동_코드_명"
]


# YYYYQ -> YYYYMM 변환
apt_price_df["YYYYMM"] = (
    apt_price_df["기준_년분기_코드"]
    .astype(str)
    .str[:4]
    +
    apt_price_df["기준_년분기_코드"]
    .astype(str)
    .str[-1]
    .map(quarter_month_map)
)

result = []

# 행정동별 월 보간
for code, g in apt_price_df.groupby("행정동코드"):

    g = g.copy()

    # 날짜 변환
    g["DATE"] = pd.to_datetime(
        g["YYYYMM"],
        format="%Y%m"
    )

    g = g.sort_values("DATE")

    # index 설정
    g = g.set_index("DATE")

    # 월 단위 확장
    monthly = g.resample("MS").asfreq()

    # 수치 컬럼 보간
    monthly[value_cols] = (
        monthly[value_cols]
        .interpolate(method="linear")
    )

    # code 채우기
    monthly[code_cols] = (
        monthly[code_cols]
        .ffill()
    )

    # 행정동코드 유지
    monthly["행정동코드"] = code

    # YYYYMM 생성
    monthly["YYYYMM"] = (
        monthly.index.strftime("%Y%m")
    )

    result.append(monthly)


# 합치기
monthly_apartment_df = (
    pd.concat(result)
    .reset_index(drop=True)
)


monthly_apartment_df[value_cols] = (
    monthly_apartment_df[value_cols]
    .round()
    .fillna(0)
    .astype(int)

)

monthly_apartment_df[code_cols] = (
    monthly_apartment_df[code_cols]
    .ffill()
)

monthly_apartment_df = (
    monthly_apartment_df
    .sort_values(
        ["YYYYMM", "행정동코드"]
    )
    .reset_index(drop=True)
)

monthly_apartment_df[[
    "기준_년분기_코드",
    "YYYYMM",
    "행정동코드",
    "행정동_코드_명",
    "아파트_단지_수",
    "아파트_면적_66_제곱미터_미만_세대_수",
    "아파트_면적_66_제곱미터_세대_수",
    "아파트_면적_99_제곱미터_세대_수",
    "아파트_면적_132_제곱미터_세대_수",
    "아파트_면적_165_제곱미터_세대_수",
    "아파트_평균_면적",
    "아파트_평균_시가"]].to_csv(
    "서울시_행정동_아파트가격_월단위_202404_202512.csv",
    index=False,
    encoding="utf-8-sig")


print(f"갯수: {len(monthly_apartment_df)}")
monthly_apartment_df.head(2)

```
